```python
# armstrong number//// 
# (52)= (5x5)+(2x2)!= 52 so it is not a armstrong number
# (123)= (1x1x1)+(2x2x2)+(3x3x3)!=123  so it is not a armstrong number

num = int (input('enter a number here:'))
order=len(str(num))
sum=0
temp=num
while temp>0:
    digit=temp%10
    cube=digit**order
    sum=sum+cube
    temp//=10

if sum==num:
      print(num,' is an armstrong number')
    
else:
      print(num,' is not an armstrong number')
```

```python
def  armstrong_num_range():

    lower=int(input('enetr the lower limit number:'))
    upper=int(input('enetr the upper limit number:'))

    for num in range(lower, upper+1):
        order=len(str(num))
        sum=0
        temp=num
        while temp >0:
            digit= temp%10
            sum+= digit ** order
            temp//=10
        if sum==num:
            print(num)


```
